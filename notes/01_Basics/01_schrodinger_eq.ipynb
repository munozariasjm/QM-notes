# Schrodinger's Equation

Most of QM is based on understanding the distributioof probabilities, which, for simple problems is basically reductible to simply understanding how to solve the ODE:

$$
\hat H\psi = E\psi\qquad\text{with}\quad H = \frac{p^2}{2m} - V(x), \quad p = -i\hbar \frac{\partial}{\partial x}
$$

$$
\left(\frac{\hbar}{2m}\partial^2/\partial x^2 -  V(x)\right)\psi = E\psi
$$
